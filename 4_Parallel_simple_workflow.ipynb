```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class MyState(TypedDict):
    runs: int
    balls: int
    sixes: int
    fours: int
    sr: float
    boundary_percentage: float
    bpd: float
    summary: str
```

```python
def calculate_sr(state:MyState)->MyState:
    sr = (state['runs']/state['balls'])*100
    return {'sr': sr}
```

```python
def calculate_boundary_percentage(state:MyState)->MyState:
    boundary_percentage = ((state['sixes'] + state['fours'])/state['balls'])*100
    return {'boundary_percentage': boundary_percentage}
```

```python
def calculate_bpd(state:MyState)->MyState:
    bpd = (state['sixes'] + state['fours'])/state['balls']
    return {'bpd': bpd}
```

```python
def generate_summary(state:MyState)->MyState:
    summary = f"Player scored {state['runs']} runs in {state['balls']} balls with a strike rate of {state['sr']:.2f}. The boundary percentage is {state['boundary_percentage']:.2f}% and boundaries per delivery is {state['bpd']:.2f}."
    return {'summary': summary}
```

```python
graph= StateGraph(MyState)
graph.add_node("calcuate_sr",calculate_sr)
graph.add_node("calculate_boundary_percentage",calculate_boundary_percentage)
graph.add_node("calculate_bpd",calculate_bpd)
graph.add_node("generate_summary",generate_summary)
graph.add_edge(START,"calcuate_sr")
graph.add_edge(START,"calculate_boundary_percentage")
graph.add_edge(START,"calculate_bpd")
graph.add_edge("calcuate_sr","generate_summary")
graph.add_edge("calculate_boundary_percentage","generate_summary")
graph.add_edge("calculate_bpd","generate_summary")
graph.add_edge("generate_summary",END)
workflow=graph.compile()
```

```python
initial_state={
    'runs': 100,
    'balls': 60,
    'sixes': 5,
    'fours': 10}
final_state=workflow.invoke(initial_state)
print(final_state)
```
